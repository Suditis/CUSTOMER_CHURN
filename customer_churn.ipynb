```python
import pandas as pd
```

```python
df = pd.read_csv("customer_churn.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
df['joining_date'] = pd.to_datetime(df['joining_date'], format='%d-%m-%Y')
```

```python
df['last_visit_time'] = pd.to_datetime(df['last_visit_time'], format='%H:%M:%S').dt.time
```

```python
df['avg_frequency_login_days'].unique()
```

```python
df['avg_frequency_login_days'] = pd.to_numeric(df['avg_frequency_login_days'], errors='coerce')
```

```python
df.drop(['security_no', 'referral_id'], axis=1, inplace=True)
```

```python
df['churn_risk_score']
```

```python
df['churn_risk_score'].dtype
```

```python
df['churn_risk_score'].value_counts()
```

```python
df['tenure_days'] = (pd.Timestamp.today() - df['joining_date']).dt.days
```

```python
df['joining_month'] = df['joining_date'].dt.month
df['joining_year'] = df['joining_date'].dt.year
```

```python
df['used_special_discount'] = df['used_special_discount'].map({'Yes':1, 'No':0})
```

```python
df.isnull().sum()
```

```python
df = df.sort_values('joining_date')
df['region_category'] = df['region_category'].bfill()
```

```python
df['avg_frequency_login_days'].fillna(
    df['avg_frequency_login_days'].median(),
    inplace=True
)
```

```python
df.isnull().sum()
```

```python
df['preferred_offer_types'] = df['preferred_offer_types'].fillna('Unknown')

df['medium_of_operation'] = df['medium_of_operation'].fillna('Unknown')
```

```python
df.isnull().sum()
```

```python
df['joined_through_referral'] = df['joined_through_referral'].replace('?', 'No')
```

```python
df['joined_through_referral'].value_counts()
```

```python
df.to_csv("cleaned_churn_data.csv", index=False)
```
